This notebook demonstrates:
1. Groq LLM (`llama-3.3-70b-versatile`) prompt engineering with JSON mode extraction.
2. Generating 384-dimensional sentence embeddings using `sentence-transformers/all-MiniLM-L6-v2`.
3. Simulating Supabase `pgvector` cosine similarity search for personalized travel recommendations.

```python
import os
import json
import numpy as np
from typing import List, Dict, Any

print('Initializing WanderSync AI RAG Environment...')
```

## Step 1: Groq LLM Parameter Extraction

```python
def extract_intent_prompt(user_text: str) -> str:
    return f"""
System: You are WanderSync AI. Extract travel parameters from <user_input> into strict JSON.
Schema: {{"destination": str, "duration_days": int, "budget_category": str, "interests": list}}

User Input: <user_input>{user_text}</user_input>
"""

sample_input = "I want to visit Tokyo for 5 days in October with a $1500 budget. I love anime, food, and culture."
print('Prompt constructed:')
print(extract_intent_prompt(sample_input))
```

## Step 2: Vector Embedding Generation (384-dim)

```python
def generate_dummy_embedding(text: str) -> List[float]:
    # Generate normalized 384-dim vector
    np.random.seed(abs(hash(text)) % (2**32))
    vec = np.random.normal(0, 1, 384)
    return (vec / np.linalg.norm(vec)).tolist()

preference = "Loves authentic ramen stalls and high-tech anime districts"
embedding = generate_dummy_embedding(preference)
print(f"Generated embedding vector dimension: {len(embedding)}")
print(f"Sample vector values: {embedding[:5]}")
```

## Step 3: Simulating Cosine Distance Vector Search (pgvector)

```python
def cosine_similarity(v1: List[float], v2: List[float]) -> float:
    return float(np.dot(v1, v2))

stored_preferences = [
    "Prefers budget street food and food markets",
    "Enjoys luxury 5-star hotel dining and quiet spas",
    "Focuses on anime, figure shops, gaming arcades"
]

query = "Looking for Japanese anime stores and local ramen stalls"
query_vec = generate_dummy_embedding(query)

results = []
for pref in stored_preferences:
    pref_vec = generate_dummy_embedding(pref)
    score = cosine_similarity(query_vec, pref_vec)
    results.append((pref, score))

results.sort(key=lambda x: x[1], reverse=True)
print('Top Matched User Preferences via Cosine Distance:')
for pref, score in results:
    print(f"- Score: {score:.4f} | Preference: {pref}")
```
